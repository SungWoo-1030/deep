# 이미지 분류 학습 및 테스트

딥러닝 앱 소스와 실행 노트북은 `Deep` 폴더에 있습니다. 앱에서 학습 폴더와 테스트 폴더 경로를 직접 입력합니다. 폴더를 지정하고 실행하기 전에는 데이터 이미지를 미리 불러오지 않습니다.

- 기본 학습 폴더는 `D:\\VisionTest\\data\\neu\\train`, 기본 테스트 폴더는 `D:\\VisionTest\\data\\neu\\test\\new`입니다. 앱의 경로 입력란에서 다른 경로로 변경할 수 있습니다.
- 데이터는 `학습폴더\\클래스명\\이미지파일` 구조로 둡니다. 클래스 폴더 이름이 분류 라벨이며, 각 이미지의 결함 영역은 학습 전에 브러시로 직접 표시합니다.
- 테스트 이미지는 별도 폴더에 두고 테스트 폴더 경로를 입력합니다. 하위 폴더도 검색하며, 학습 폴더에 없던 클래스는 미리 예측할 수 없습니다.
- 라벨링 탭에서 폴더를 불러오면 클래스별 이미지 목록에 각 사진의 **라벨 상태**와 **현재 모델 학습 반영 상태**가 표시됩니다. 목록 행을 클릭하면 해당 사진이 열립니다. 마스크를 저장하면 상태가 즉시 갱신됩니다.
- 각 학습 이미지를 브러시로 표시하고 저장해야 학습할 수 있습니다. 정상 이미지는 `good` 또는 `normal` 클래스 폴더에 두고 빈 마스크를 저장합니다.
- 학습 버튼을 누르면 이미지 종류 분류와 결함 픽셀 영역 분할을 함께 학습하고 모델을 저장합니다.
- 테스트 결과는 표로 표시됩니다. 표의 이미지 행을 클릭하면 해당 사진과 모델이 예측한 결함 영역을 아래에서 확인할 수 있습니다. 빨간색 예측 영역은 정답 보장이 아니며, 예측 마스크는 `Deep/outputs/predicted_masks`에 저장됩니다.
- 분류 결과는 `Deep/outputs/test_predictions.csv`에 저장됩니다.
- 결과 점수는 보정된 확률이나 정확도 보증이 아닙니다.
